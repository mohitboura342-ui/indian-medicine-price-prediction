````python
price_col = "price(₹)"
match_cols = [
    "short_composition1",
    "short_composition2",
    "pack_size_label"
]

# Comparison ke liye text values normalize karein
for col in match_cols:
    df[col + "_match"] = (
        df[col].astype("string").str.strip().str.lower().fillna("[missing]")
    )

group_cols = [col + "_match" for col in match_cols]

df["peer_count"] = df.groupby(group_cols)[price_col].transform("size")
df["peer_median_price"] = df.groupby(group_cols)[price_col].transform("median")

# ₹0 price ya peer median ke aadhe se kam price ko review ke liye flag karein
df["review_flag"] = (
    (df[price_col] <= 0)
    | (
        (df[price_col] < 10)
        & (df["peer_count"] >= 5)
        & (df[price_col] < 0.5 * df["peer_median_price"])
    )
)

review_rows = df[df["review_flag"]]

review_rows[
    [
        "name",
        "price(₹)",
        "pack_size_label",
        "short_composition1",
        "short_composition2",
        "peer_count",
        "peer_median_price",
    ]
]
````

- Prices were compared among medicines with the same composition and pack size. Records priced below ₹10 and less than half their peer-group median were flagged for review. A flag does not mean the price is incorrect; it only indicates that it should be verified.

## Column Added or Modified

- **`short_composition2`** — No new column was added. Missing values in this existing column were replaced with **“No second composition listed”** to make the data clearer and easier to analyze.

- ## Additional Columns for Price Comparison

These columns were added to compare each medicine with similar products and flag prices for review:

- **`short_composition1_match`** — Normalized first listed active ingredient, used to match medicines with the same composition.
- **`short_composition2_match`** — Normalized second listed active ingredient; missing values are treated as “no second composition listed” for matching.
- **`pack_size_label_match`** — Normalized pack size, used to compare medicines sold in similar quantities or forms.
- **`peer_count`** — Number of records with the same composition and pack size. It helps show how many comparable medicines are available.
- **`peer_median_price`** — Median price of those comparable records; used as a reference price.
- **`review_flag`** — Marks records with ₹0 price or a price below ₹10 that is less than half the peer median, when at least five comparable records exist. A flag means “review this record,” not “this price is wrong.”

Together, these columns help identify prices that differ from comparable medicines, while keeping the original price values unchanged.

## Data-Cleaning Decision

These comparisons support the plausibility of the high prices examined, but they do not verify every record or establish when the dataset’s prices were recorded. Therefore, high-priced medicines have been retained. A record should be removed only if a check of the same medicine, strength, pack size, and relevant price date confirms an error.

> **Note:** High-priced medicines were retained because specialty and cancer medicines can be costly in India, and a boxplot outlier does not automatically mean the price is wrong.

````python
import pandas as pd
from IPython.display import display

column_summary = []

for col in df.columns:
    values = df[col].dropna().unique()
    counts = df[col].value_counts(dropna=False)

    # Kam unique values wale columns ke sab values dikhayein;
    # high-cardinality columns mein pehle 10 examples dikhayein.
    if len(values) <= 20:
        unique_values_text = ", ".join(map(str, values))
    else:
        unique_values_text = (
            ", ".join(map(str, values[:10])) + " ... (more values)"
        )

    top_values_text = "; ".join(
        f"{value}: {count}"
        for value, count in counts.head(10).items()
    )

    row = {
        "column": col,
        "data_type": str(df[col].dtype),
        "missing_values": int(df[col].isna().sum()),
        "missing_percent": round(df[col].isna().mean() * 100, 2),
        "unique_value_count": df[col].nunique(dropna=True),
        "unique_values_or_sample": unique_values_text,
        "top_10_values_with_counts": top_values_text
    }

    if pd.api.types.is_numeric_dtype(df[col]):
        row["minimum"] = df[col].min()
        row["median"] = df[col].median()
        row["mean"] = round(df[col].mean(), 2)
        row["maximum"] = df[col].max()

    column_summary.append(row)

column_summary_df = pd.DataFrame(column_summary)
display(column_summary_df)
````

````python
price = df["price(₹)"].dropna()

bins = [-1, 10, 50, 100, 500, 1_000, 5_000, 10_000,
        50_000, 100_000, float("inf")]

labels = [
    "₹0–<₹10", "₹10–<₹50", "₹50–<₹100", "₹100–<₹500",
    "₹500–<₹1k", "₹1k–<₹5k", "₹5k–<₹10k",
    "₹10k–<₹50k", "₹50k–<₹1 lakh", "₹1 lakh+"
]

price_ranges = pd.cut(price, bins=bins, labels=labels, right=False)
counts = price_ranges.value_counts(sort=False)
percentages = counts / len(price) * 100

# Graph ke liye automatic takeaway
most_common_range = counts.idxmax()
most_common_percent = percentages.loc[most_common_range]
high_price_percent = (price >= 10_000).mean() * 100

takeaway = (
    f"Takeaway: Most medicines are in {most_common_range} "
    f"({most_common_percent:.1f}%); "
    f"{high_price_percent:.1f}% are priced at ₹10,000 or more."
)

fig, ax = plt.subplots(figsize=(12, 6))

sns.barplot(
    x=counts.index.astype(str),
    y=counts.values,
    color="steelblue",
    ax=ax
)

for bar, count, percent in zip(ax.patches, counts, percentages):
    ax.annotate(
        f"{count:,}\n({percent:.1f}%)",
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        ha="center",
        va="bottom",
        fontsize=9
    )

fig.suptitle("Number of Medicines by Price Range", fontsize=15)
fig.text(
    0.5, 0.90, takeaway,
    ha="center",
    bbox=dict(facecolor="lightyellow", edgecolor="gray", boxstyle="round,pad=0.5")
)

ax.set_xlabel("Price Range (₹)")
ax.set_ylabel("Number of Medicines")
ax.tick_params(axis="x", rotation=35)
plt.subplots_adjust(top=0.78, bottom=0.25)
plt.show()
````

### Listed Prices and High-Cost Medicines

Prices in this chart range from **₹3,500 to ₹1,19,500** among the selected products; this is a comparison of the three highest-priced entries for each of the ten most represented manufacturers, not a company-wide price range. For example, an [NPPA list](https://nppa.gov.in/uploads/pdf/List-of-526-brandspdf-bede2ef110008d6c2aa00ecf1708f960.pdf) gives a revised MRP for **Bryxta 400 mg/16 ml** that differs from the chart.

````python
import pandas as pd
import matplotlib.pyplot as plt
import textwrap

price_col = "price(₹)"
comp1_col = "short_composition1"
comp2_col = "short_composition2"
pack_col = "pack_size_label"

# Relevant columns ki copy aur cleaning
data = df[[price_col, comp1_col, comp2_col, pack_col]].copy()
data[price_col] = pd.to_numeric(data[price_col], errors="coerce")
data = data.dropna(subset=[price_col])
data = data[data[price_col] >= 0]

for col in [comp1_col, comp2_col, pack_col]:
    data[col] = data[col].fillna("Not listed").astype(str).str.strip()
    data[col] = data[col].replace("", "Not listed")

# Exact peer group: same first/second composition and same pack-size label
data["Composition + pack"] = (
    data[comp1_col] + " | " + data[comp2_col] + " | " + data[pack_col]
)

# Sirf groups jahan comparison ke liye kam-se-kam 30 records hon
minimum_records = 30

composition_summary = (
    data.groupby(comp1_col)[price_col]
    .agg(
        n="count",
        median="median",
        q1=lambda x: x.quantile(0.25),
        q3=lambda x: x.quantile(0.75)
    )
    .query("n >= @minimum_records")
    .nlargest(12, "n")
    .sort_values("median")
)

peer_summary = (
    data.groupby("Composition + pack")[price_col]
    .agg(
        n="count",
        median="median",
        q1=lambda x: x.quantile(0.25),
        q3=lambda x: x.quantile(0.75)
    )
    .query("n >= @minimum_records")
    .nlargest(12, "n")
    .sort_values("median")
)

overall_median = data[price_col].median()

fig, axes = plt.subplots(1, 2, figsize=(18, 10))

def draw_price_summary(ax, summary, title, wrap_width=32):
    if summary.empty:
        ax.set_title(title + "\n(No groups meet the minimum record count)")
        ax.axis("off")
        return

    labels = [
        textwrap.fill(f"{label} (n={int(row['n']):,})", width=wrap_width)
        for label, row in summary.iterrows()
    ]

    y_positions = range(len(summary))

    for y, (_, row) in zip(y_positions, summary.iterrows()):
        # Line = 25th to 75th percentile; dot = median price
        ax.errorbar(
            row["median"],
            y,
            xerr=[[
                row["median"] - row["q1"]
            ], [
                row["q3"] - row["median"]
            ]],
            fmt="o",
            color="#2878B5",
            ecolor="#8AB6D6",
            elinewidth=3,
            capsize=4,
            markersize=7
        )

    ax.axvline(
        overall_median,
        color="#E07A2D",
        linestyle="--",
        linewidth=2,
        label=f"Overall median: ₹{overall_median:,.0f}"
    )

    ax.set_yticks(list(y_positions))
    ax.set_yticklabels(labels)
    ax.set_xlabel("Listed price (₹)")
    ax.set_title(title, fontsize=13, fontweight="bold")
    ax.grid(axis="x", linestyle="--", alpha=0.3)
    ax.set_axisbelow(True)
    ax.legend(loc="best")
    ax.spines[["top", "right", "left"]].set_visible(False)

draw_price_summary(
    axes[0],
    composition_summary,
    "Price by First Composition\n(12 most frequent eligible compositions)"
)

draw_price_summary(
    axes[1],
    peer_summary,
    "Price within Same Composition + Pack-Size Groups\n(12 most frequent eligible groups)",
    wrap_width=42
)

fig.suptitle(
    "How Listed Medicine Prices Vary by Composition and Pack Size",
    fontsize=17,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

print(f"Overall median listed price: ₹{overall_median:,.2f}")
print(f"Composition groups shown: {len(composition_summary)}")
print(f"Same-composition-and-pack groups shown: {len(peer_summary)}")
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import textwrap

# Price ko numeric banao
df["price(₹)"] = pd.to_numeric(df["price(₹)"], errors="coerce")

group_columns = [
    "short_composition1_match",
    "short_composition2_match",
    "pack_size_label_match"
]

# Teeno comparison columns aur price available hone wale records lo
data = df.dropna(subset=group_columns + ["price(₹)"]).copy()

# Same ingredients + same pack size ke groups ka summary
price_comparison = (
    data.groupby(group_columns, dropna=False)
        .agg(
            records=("price(₹)", "size"),
            median_price=("price(₹)", "median"),
            q1=("price(₹)", lambda x: x.quantile(0.25)),
            q3=("price(₹)", lambda x: x.quantile(0.75)),
            lowest_price=("price(₹)", "min"),
            highest_price=("price(₹)", "max"),
            example_medicine=("name", "first")
        )
        .reset_index()
)

# Bahut chhote groups ka comparison unstable ho sakta hai
min_records = 10
plot_data = price_comparison[
    price_comparison["records"] >= min_records
].copy()

if plot_data.empty:
    print(f"{min_records} ya usse zyada records wale matching groups nahi mile.")
else:
    # Middle 50% prices ke spread ko calculate karo
    plot_data["iqr"] = plot_data["q3"] - plot_data["q1"]

    # Sabse zyada IQR wale 12 groups dikhayein
    plot_data = plot_data.nlargest(12, "iqr").sort_values("median_price")

    labels = (
        plot_data["short_composition1_match"].astype(str)
        + " + "
        + plot_data["short_composition2_match"].astype(str)
        + "\n"
        + plot_data["pack_size_label_match"].astype(str)
    )

    labels = [textwrap.fill(label, width=48) for label in labels]

    fig, ax = plt.subplots(figsize=(14, 9))
    y_positions = range(len(plot_data))

    # Q1 se Q3 tak line: beech ke 50% prices
    for y, (_, row) in zip(y_positions, plot_data.iterrows()):
        ax.hlines(
            y=y,
            xmin=row["q1"],
            xmax=row["q3"],
            color="steelblue",
            linewidth=5,
            alpha=0.8
        )

        # Median price
        ax.scatter(
            row["median_price"],
            y,
            color="darkorange",
            s=75,
            zorder=3
        )

        # Har group ka record count
        ax.text(
            row["q3"],
            y + 0.15,
            f'n={int(row["records"]):,}',
            fontsize=9
        )

    ax.set_yticks(list(y_positions))
    ax.set_yticklabels(labels)
    ax.set_xlabel("Listed price (₹)")
    ax.set_ylabel("Same composition and pack size")
    ax.set_title(
        "Price Spread Among Comparable Medicine Records",
        fontsize=15,
        weight="bold"
    )

    ax.scatter([], [], color="darkorange", s=75, label="Group median price")
    ax.plot([], [], color="steelblue", linewidth=5, label="Middle 50% of prices (Q1–Q3)")
    ax.legend(loc="best")
    ax.grid(axis="x", linestyle="--", alpha=0.3)

    plt.tight_layout()
    plt.show()
display(
    plot_data[
        group_columns
        + [
            "records",
            "median_price",
            "q1",
            "q3",
            "iqr",
            "lowest_price",
            "highest_price",
            "example_medicine",
        ]
    ].sort_values("iqr", ascending=False)
)



   
````

## Understanding Price Differences in Comparable Medicines

Medicines with the same listed composition and pack size can still have different listed prices. For example, abiraterone acetate 250 mg records range from **₹13,000 to ₹1,50,000**, with a median of **₹30,000**—showing why students should compare the **exact brand, manufacturer, formulation, pack, and price date** before drawing conclusions.  
A wide price range is a prompt for further investigation; it does not, by itself, prove overpricing or explain the reason for the difference. The second composition is missing in these records, so it could not be included in the comparison.  
For an official India price check, search the medicine’s exact details on [NPPA Pharma Sahi Daam](https://nppaipdms.gov.in/NPPA/PharmaSahiDaam/searchMedicine).


````python
import pandas as pd
import matplotlib.pyplot as plt
import math
import textwrap

# Required columns check karo
required_columns = [
    "price(₹)",
    "pack_size_label",
    "short_composition1_match",
    "short_composition2_match",
]

missing = [col for col in required_columns if col not in df.columns]

if missing:
    print("DataFrame mein yeh columns nahi mile:", missing)

else:
    data = df.copy()

    # Price ko numeric banao
    data["price(₹)"] = pd.to_numeric(data["price(₹)"], errors="coerce")

    # Pack label se tablet/capsule quantity aur form nikalo
    pack_text = data["pack_size_label"].fillna("").astype(str).str.lower()

    pack_info = pack_text.str.extract(
        r"(\d+(?:\.\d+)?)\s*(tablets?|capsules?)"
    )

    data["pack_quantity_num"] = pd.to_numeric(
        pack_info[0], errors="coerce"
    )

    data["dosage_form_for_graph"] = (
        pack_info[1]
        .str.lower()
        .str.replace(r"s$", "", regex=True)
    )

    # Missing composition ko ek consistent label do
    for col in ["short_composition1_match", "short_composition2_match"]:
        data[col] = (
            data[col]
            .fillna("Not listed")
            .astype(str)
            .str.strip()
        )

    # Valid tablet/capsule records rakho
    data = data.dropna(
        subset=[
            "price(₹)",
            "pack_quantity_num",
            "dosage_form_for_graph",
        ]
    ).copy()

    # Zero ya negative quantity ko hatao
    data = data[data["pack_quantity_num"] > 0].copy()

    if data.empty:
        print("Pack-size label se tablet/capsule quantity nahi mil payi.")

    else:
        # Per tablet/capsule listed price
        data["price_per_unit"] = (
            data["price(₹)"] / data["pack_quantity_num"]
        )

        composition_keys = [
            "short_composition1_match",
            "short_composition2_match",
            "dosage_form_for_graph",
        ]

        group_columns = composition_keys + ["pack_quantity_num"]

        # Same ingredients, form aur pack quantity ka summary
        summary = (
            data.groupby(group_columns, dropna=False)
            .agg(
                records=("price(₹)", "size"),
                median_pack_price=("price(₹)", "median"),
                median_price_per_unit=("price_per_unit", "median"),
            )
            .reset_index()
        )

        # Har pack-size group mein kam-se-kam 3 records chahiye
        summary = summary[summary["records"] >= 3].copy()

        if summary.empty:
            print(
                "Kisi pack-size group mein 3 ya zyada records nahi mile. "
                "Code mein records >= 3 ko records >= 1 karke dobara try kar sakte ho."
            )

        else:
            # Aise ingredient/form groups dhoondo jahan 2+ pack sizes hon
            eligible = (
                summary.groupby(composition_keys, dropna=False)
                .agg(
                    pack_sizes=("pack_quantity_num", "nunique"),
                    total_records=("records", "sum"),
                )
                .reset_index()
            )

            eligible = eligible[eligible["pack_sizes"] >= 2]

            if eligible.empty:
                print(
                    "Same composition aur form ke 2 alag pack sizes "
                    "compare karne layak nahi mile."
                )

            else:
                # Sabse zyada records wale top 6 groups
                selected = eligible.sort_values(
                    "total_records", ascending=False
                ).head(6)

                plot_data = summary.merge(
                    selected[composition_keys],
                    on=composition_keys,
                    how="inner",
                )

                n_plots = len(selected)
                ncols = 2
                nrows = math.ceil(n_plots / ncols)

                fig, axes = plt.subplots(
                    nrows,
                    ncols,
                    figsize=(14, 5 * nrows),
                    squeeze=False,
                )
                axes = axes.flatten()

                for ax, (_, group) in zip(axes, selected.iterrows()):
                    mask = pd.Series(True, index=plot_data.index)

                    for col in composition_keys:
                        mask &= plot_data[col].eq(group[col])

                    subset = plot_data.loc[mask].sort_values(
                        "pack_quantity_num"
                    )

                    ax.plot(
                        subset["pack_quantity_num"],
                        subset["median_price_per_unit"],
                        marker="o",
                        linewidth=2,
                        color="teal",
                    )

                    for _, row in subset.iterrows():
                        ax.annotate(
                            f'n={int(row["records"])}',
                            (
                                row["pack_quantity_num"],
                                row["median_price_per_unit"],
                            ),
                            xytext=(5, 6),
                            textcoords="offset points",
                            fontsize=8,
                        )

                    form = str(group["dosage_form_for_graph"])

                    title = (
                        f'{group["short_composition1_match"]} + '
                        f'{group["short_composition2_match"]} | '
                        f'{form.title()}'
                    )

                    ax.set_title(textwrap.fill(title, 42), fontsize=10)
                    ax.set_xlabel(f"Pack quantity ({form}s)")
                    ax.set_ylabel("Median listed price per unit (₹)")
                    ax.grid(True, linestyle="--", alpha=0.3)

                # Extra khaali panels hide karo
                for ax in axes[n_plots:]:
                    ax.set_visible(False)

                fig.suptitle(
                    "Pack Size and Median Price per Tablet/Capsule",
                    fontsize=15,
                    fontweight="bold",
                )

                plt.tight_layout()
                plt.show()

                # Graph ke numbers ki summary table
                display(
                    plot_data[
                        composition_keys
                        + [
                            "pack_quantity_num",
                            "records",
                            "median_pack_price",
                            "median_price_per_unit",
                        ]
                    ]
                    .sort_values(
                        composition_keys + ["pack_quantity_num"]
                    )
                    .reset_index(drop=True)
                )
````

## 3. Foundation Chapter: Basic Principles of Pathophysiology

>  **Pathophysiology = Physiology (normal function) + Pathology (abnormal change).** It explains *how* normal body function breaks down in disease, and therefore *where* a drug can step in.

### 3.1 Cell injury and adaptation

```text
Injury (hypoxia, toxins, microbes, trauma, immune reactions)
            ↓
Reversible injury (cell swelling, fatty change)  ← cell recovers if cause is removed
            ↓  (cause persists)
Irreversible injury → cell death
            ↓
Necrosis (accidental, causes inflammation)  |  Apoptosis (programmed, no inflammation)
```

| Adaptation | Meaning | Example |
|:--|:--|:--|
| Hypertrophy | Cells become larger | Heart muscle in long-standing hypertension |
| Hyperplasia | Cell number increases | Endometrium under hormonal stimulation |
| Atrophy | Cells shrink | Disuse of muscle |
| Metaplasia | One cell type replaced by another | Bronchial lining in smokers |
| Dysplasia | Disordered, pre-cancerous growth | Cervical epithelium |

### 3.2 Inflammation, the most important basic process

```text
Injury / infection
      ↓
Vascular changes: vasodilation + increased permeability (histamine, prostaglandins)
      ↓
Cellular events: neutrophils and macrophages arrive (chemotaxis, phagocytosis)
      ↓
Cardinal signs: Rubor (redness), Calor (heat), Tumor (swelling), Dolor (pain), Functio laesa (loss of function)
```

| Mediator | Role | Medicines in your dataset that block it |
|:--|:--|:--|
| Histamine | Itching, swelling, runny nose | Fexofenadine, Levocetirizine, Pheniramine, Chlorpheniramine |
| Prostaglandins (via COX) | Pain, fever, swelling | Aceclofenac, Diclofenac, Nimesulide, Piroxicam, Mefenamic acid |
| Leukotrienes | Airway constriction, mucus | Montelukast |
| Broad inflammatory pathways | Chronic inflammation | Deflazacort (corticosteroid) |

>  **NSAID** → blocks COX → less prostaglandin → less pain, fever and swelling. **Antihistamine** → blocks H1 receptor → fewer allergy symptoms.

### 3.3 Hypersensitivity (allergy)

```text
Allergen → IgE binds mast cells → degranulation → histamine + leukotrienes
        → sneezing, itching, wheezing, hives
```
Type I hypersensitivity underlies allergic rhinitis, urticaria and allergic asthma.

### 3.4 Neoplasia (cancer) hallmarks

Uncontrolled proliferation • escape from apoptosis • new blood vessel formation (angiogenesis) • invasion and metastasis • genetic instability. Each hallmark is a drug target (see Chapter G).

**What goes wrong:** Microbes enter, multiply and trigger inflammation, fever and tissue damage.

```text
Microbe entry → multiplication → immune response → inflammatory mediators → fever, pain, discharge
```

| Ornidazole, Metronidazole | Nitroimidazoles | DNA damage in anaerobes and protozoa |
| Albendazole | Anthelmintic | Worm microtubules |
| Amphotericin B | Antifungal | Binds fungal ergosterol |

**What goes wrong:** After injury, prostaglandins sensitise nociceptors and amplify pain and swelling.

```text
Tissue injury → COX enzyme activated → prostaglandin ↑ → nociceptor sensitisation → pain + swelling + fever
```

### Chapter C: Allergy and respiratory diseases

**Normal:** Open airways, controlled mucus, balanced immunity.
**What goes wrong:** Allergen → histamine / leukotriene release → bronchoconstriction, mucus, swelling.

```text
Allergen → IgE → mast-cell degranulation → histamine + leukotrienes
      ↓
Airway smooth-muscle contraction + mucus + oedema → wheeze, cough, breathlessness
```

| Problem | Drug class | Examples in dataset |
|:--|:--|:--|
| Allergic symptoms | H1 antihistamines | Fexofenadine, (Levo)cetirizine, Pheniramine, Chlorpheniramine, Hydroxyzine |
| Airway inflammation | Leukotriene antagonist | Montelukast 10 mg |
| Bronchospasm | β2 agonists | Salbutamol, Levosalbutamol |
| Thick mucus | Mucolytic | Ambroxol |
| Nasal congestion | α1 agonist decongestant | Phenylephrine |

**Examples:** Allegra 120 (Fexofenadine), Allegra-M (Montelukast + Fexofenadine), Ascoril LS (Ambroxol + Levosalbutamol).

**Data link:** Levocetirizine 5 mg is the top allergy ingredient (4,248 records). Levocetirizine + Montelukast (strip of 10) has 2,237 records at a median of about Rs 103.

>  Three different targets: **cetirizine = H1 receptor, montelukast = leukotriene receptor, salbutamol = β2 receptor.** Different targets are exactly why combinations make pharmacological sense.

### Chapter D: Gastrointestinal diseases (acidity, GERD, ulcer, nausea)

**Normal:** Acid digests food; mucus and bicarbonate protect the stomach lining.
**What goes wrong:** Aggressive factors (acid, NSAIDs, *H. pylori*) outweigh protective factors, or the lower oesophageal sphincter weakens.

```text
Excess acid / NSAIDs / H. pylori / weak LES → mucosal injury → ulcer or reflux → burning, pain, nausea
```

| Drug | Class | Mechanism |
|:--|:--|:--|
| Pantoprazole, Rabeprazole, Omeprazole, Esomeprazole, Lansoprazole | Proton pump inhibitors | Block H⁺/K⁺-ATPase → less acid |
| Domperidone | Prokinetic / antiemetic | D2 blockade → faster gastric emptying, less nausea |
| Aluminium / magnesium hydroxide | Antacids | Neutralise acid |

**Data link:** Domperidone 30 mg (5,133 records) and 10 mg (3,271); Pantoprazole 40 mg (2,707 as main + 2,896 as partner); Rabeprazole 20 mg (2,423 as main + 4,743 as partner). Domperidone 30 + Rabeprazole 20 capsule has about 2,000–2,400 records at a median near Rs 87.

> PPI = *reduces acid*. Domperidone = *improves motility and stops vomiting*. Together they treat reflux with nausea.

---

### Chapter E: Diabetes mellitus

**Normal:** Food → blood glucose rises → β-cells release insulin → muscle and fat take up glucose, liver reduces glucose output.
**What goes wrong:** Insulin resistance and/or β-cell failure.

```text
Insulin resistance / β-cell dysfunction
      ↓
Less glucose uptake + more liver glucose output
      ↓
Hyperglycaemia → damage to eyes, kidneys, nerves, heart over time
```

| Drug | Class | Mechanism |
|:--|:--|:--|
| Metformin | Biguanide | ↓ liver glucose output, ↑ insulin sensitivity |
| Glimepiride, Gliclazide | Sulfonylureas | Close β-cell K-ATP channels → ↑ insulin release |
| Sitagliptin | DPP-4 inhibitor | ↑ incretins → glucose-dependent insulin release |

**Data link:** Glimepiride appears in two top-15 strengths (2 mg: 2,520 and 1 mg: 2,256 = 4,776 records). Metformin 500 mg appears as a partner in 4,532 records. These two numbers are strikingly close.

> Sulfonylureas can cause **hypoglycaemia** because they force insulin release; metformin rarely does.

### Chapter F: Hypertension and cardiovascular disease

**Normal:** **Blood pressure = cardiac output × peripheral resistance**, regulated by the kidney, RAAS and the nervous system.
**What goes wrong:** Vasoconstriction, salt and water retention and neurohormonal activation raise pressure persistently.

```text
Angiotensin II ↑ / vasoconstriction ↑ / Na⁺ + water retention
      ↓
Persistently high BP → cardiac hypertrophy, vessel damage → stroke, MI, kidney damage
```

| Drug | Class | Mechanism |
|:--|:--|:--|
| Telmisartan, Losartan, Olmesartan | ARBs | Block AT1 receptor → vasodilation, ↓ aldosterone |
| Amlodipine | Calcium channel blocker | Blocks L-type Ca²⁺ channels → vessel relaxation |

**Data link:** Telmisartan 40 mg has 3,259 records (median about Rs 82), the only blood-pressure drug in the top-15.

> 💡 ARB side effect: hyperkalaemia. Amlodipine side effect: ankle oedema.

```text
Genetic / epigenetic changes → loss of growth control → tumour → invasion → metastasis
```

## 7. Deep-Dive Insights from the Notebook

This section connects the numbers to pathophysiology and pharmacology. Statements marked **(inference)** are reasoned interpretations of the data rather than facts printed in the notebook.

### 7.1 A tiny group of ingredients covers a large share of the market

There are 7,429 distinct main-ingredient values, yet the 15 shown in Graph 6 account for **54,555 records, about 21.5% of the whole dataset**. In other words, about 0.2% of ingredient values cover roughly one in five records.

**Why:** Common, high-burden conditions (infection, pain, acidity, allergy, diabetes, blood pressure) need many strengths, brands and formulations. **(inference)**

### 7.2 Acute, symptom-driven diseases lead the top-15

Infection (28%), acidity/GI (25%) and pain/inflammation (24%) together make up about **77%** of the top-15 volume. Chronic diseases such as diabetes (9%) and hypertension (6%) are present but smaller in this view.

**Pathophysiology link:** Infections, reflux and pain are frequent, recurring and symptom-driven, so many brands compete in them. **(inference)**

### 7.3 The "anchor + partner" design of combination medicines

Combination records are 44.2% of the dataset, and the partner ingredients follow clear pharmacological logic:

| Anchor drug | Partner | Biological reason |
|:--|:--|:--|
| Amoxycillin | Clavulanic acid (5,117) / Sulbactam (2,509) | The partner blocks β-lactamase, the resistance enzyme |
| Aceclofenac / Diclofenac | Paracetamol (8,897 + 3,141) | Two different pain pathways |
| Domperidone | Rabeprazole (4,743) | Less acid + better motility |
| Glimepiride | Metformin (4,532) | More insulin release + less liver glucose |
| Levocetirizine | Montelukast (2,591) | Blocks histamine and leukotriene effects |
| Ofloxacin | Ornidazole (2,396) | Covers bacteria and anaerobes/protozoa |

**Key observation:** Glimepiride (4,776 records across two strengths) and Metformin as a partner (4,532) are almost the same size, which suggests that **most glimepiride products are sold as glimepiride + metformin combinations. (inference)**

Another observation: **the same molecule plays different roles.** Rabeprazole is a main ingredient in 2,423 records and a partner in 4,743; Pantoprazole is main in 2,707 and partner in 2,896; Ofloxacin is main in 4,296 and partner in 1,664. Paracetamol, however, appears as a partner and not among the top main ingredients.

### 7.4 Price follows biology and manufacturing complexity

| Product type | Typical price signal |
|:--|:--|
| Older NSAIDs (e.g. Nimesulide) | Median about Rs 35 |
| Overall dataset | Median Rs 79 |
| Newer cephalosporin (Cefpodoxime) | Median about Rs 205 |
| Antibiotic combination (Amoxycillin + Clavulanic acid) | Rs 19.68 per tablet |
| Cancer injection / biologic (Bevacizumab 400 mg) | Median Rs 43,659 (about 550× the overall median) |

**Insights:**
- Small-molecule generics (NSAIDs, acid reducers) sit at the low end; **biologic antibodies** (bevacizumab, rituximab) sit at the top. Biologics are produced in living cells, so manufacturing is far more complex than synthesising a tablet. **(inference)**
- **Antibiotics cost the most per tablet** among the compared tablets (Rs 19.68 and Rs 23.62), compared with Rs 5–10 for pain, acidity and allergy products.
- Even inside one cancer drug the range can be enormous: **Abiraterone 250 mg runs from Rs 13,000 to Rs 1,50,000**, with a median of Rs 30,000.

### 7.5 Pack size mirrors disease duration

Azithromycin appears in packs of **3 and 5 tablets** (a short course), whereas pain, allergy and acid medicines are mostly **strips of 10**. This matches the pathophysiology: an acute bacterial infection is treated with a short, defined course, while symptomatic conditions are treated as needed. **(inference)**

Also, a larger pack is not automatically cheaper per unit, but it can be: the 15-tablet Aceclofenac + Paracetamol pack works out to Rs 5.10 per tablet versus Rs 6.90 for the 10-tablet pack, about **26% cheaper per tablet** (though based on only 24 records).

### 7.6 The company landscape is a long tail

- 7,648 companies share 253,973 records, an average of about **33 records per company**.
- The top 5 companies together hold 11,591 records, only about **4.6%** of the dataset.

This means the catalogue is spread across thousands of manufacturers rather than concentrated in a few. For students: this is why one drug (for example aceclofenac + paracetamol) can have thousands of competing brands. **(inference)**

### 7.7 Discontinuation is a company story more than a disease story

Overall, 3.1% of records are discontinued, but company rates range from **2.2% (Leeford) to 41.7% (Glenmark)**. That is a spread of nearly 19 times between companies, which suggests that portfolio decisions by individual companies drive discontinuation more than any single disease area. **(inference)**

### 7.8 Brand names are almost all unique, except commodities

98.2% of the 253,973 records have a unique brand name (249,398 unique names). The names that repeat are mainly **commodity products such as IV fluids** (NS 0.9%, Ringer Lactate), which are generic, standardised and made by multiple companies.

### 7.9 Very cheap medicines are the old, essential molecules

There are 3,432 records priced under Rs 10, including examples such as Aspirin 300 mg and a plain Antacid tablet at Rs 2. Such products are typically long-established molecules. **(inference)** In contrast, the four Rs 0 records are COVID-19 vaccines (Covishield, Covaxin, Moderna, Sputnik V), reflecting free public-health supply during the pandemic.

### 7.10 Putting it together: a story the data tells

```text
Common diseases (infection, pain, acidity, allergy, diabetes, BP)
        → many ingredients, strengths and brands in the market
Disease complexity (resistance, multiple pathways)
        → combination products (clavulanic acid, paracetamol, metformin, montelukast…)
Disease duration (acute vs chronic)
        → short antibiotic courses vs 10-tablet symptomatic strips
Therapeutic complexity (biologics, targeted cancer therapy)
        → prices hundreds of times above the median
```

> **The big idea for exams and viva:** Pathophysiology decides the drug target; the drug target decides the drug class; the drug class decides the ingredient; and the market data shows how that ingredient actually reaches patients as products, packs and prices.

> **Reading the data wisely:** These are catalogue (product) records. Record counts show how many products exist, not how many patients are treated or how effective a medicine is. Use them to understand the market and the logic of therapy, and use clinical sources for treatment decisions.

````python
df["price(₹)"] = pd.to_numeric(df["price(₹)"], errors="coerce")

df = df.dropna(subset=["price(₹)"])

X = df.drop(columns=["price(₹)"])
y = df["price(₹)"]

print("Final shape:", df.shape)
````

````python
target = "price(₹)"

features = [
    "id",
    "name",
    "Is_discontinued",
    "manufacturer_name",
    "Medicine Type",
    "pack_size_label",
    "short_composition1",
    "short_composition2",
    "short_composition1_match",
    "short_composition2_match",
    "pack_size_label_match",
    "peer_count",
    "peer_median_price",
    "review_flag",
    "pack_quantity",
    "therapeutic_category"
]

X = df[features]
y = df[target]

print("X shape:", X.shape)
print("y shape:", y.shape)
````

````python
df["Is_discontinued"] = df["Is_discontinued"].astype(int)
df["review_flag"] = df["review_flag"].astype(int)
````

````python
from sklearn.model_selection import train_test_split

X = df[features]
y = df["price(₹)"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
````

````python
categorical_features = [
    "name",
    "manufacturer_name",
    "Medicine Type",
    "pack_size_label",
    "short_composition1",
    "short_composition2",
    "short_composition1_match",
    "short_composition2_match",
    "pack_size_label_match",
    "pack_quantity",
    "therapeutic_category"
]

numerical_features = [
    "id",
    "Is_discontinued",
    "peer_count",
    "peer_median_price",
    "review_flag"
]
````

````python
for col in categorical_features:
    print(col, ":", df[col].nunique())
````

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
````

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "num",
            "passthrough",
            numerical_features
        )
    ]
)

print("Preprocessor ready!")
````

````python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

print("Random Forest model ready!")
````

````python
from sklearn.pipeline import Pipeline

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", rf_model)
    ]
)

print("Pipeline ready!")
````

````python
model.fit(X_train, y_train)

print("✅ Random Forest training completed!")
````

````python
y_pred = model.predict(X_test)

print("Prediction completed!")
print("First 10 predictions:")
print(y_pred[:10])
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)

rmse = mean_squared_error(
    y_test,
    y_pred
) ** 0.5

r2 = r2_score(y_test, y_pred)

print("Random Forest Regression Evaluation")
print("-" * 40)
print("MAE  :", mae)
print("RMSE :", rmse)
print("R²   :", r2)
````

````python
results = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred
})

results["Error"] = (
    results["Predicted Price"] - results["Actual Price"]
)

results["Absolute Error"] = abs(results["Error"])

results.head(10)
````
